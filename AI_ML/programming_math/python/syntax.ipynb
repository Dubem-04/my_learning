##### **Underscore Conventions**

- Single Leading (`_variable`): Hints at a "protected" or internal-use variable. Restricts wildcard imports.
- Double Leading (`__variable`): Triggers **name mangling** inside classes to prevent subclass naming collisions.
- Double Leading & Trailing (`__variable__`): Reserved for Python system-defined "dunder" methods or attributes.
- Single Trailing Underscore (`variable_`): Use this when your ideal variable name clashes with a built-in Python keyword.
- Single Standalone (`_`): Acts as a throwaway variable for ignored or temporary values.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance  # Internal: please don't touch this directly

account = BankAccount("Alice", 1000)

# You CAN access it, but you SHOULDN'T:
print(account._balance)
```

```python
class Parent:
    def __init__(self):
        self.__secret = "Hidden"

class Child(Parent):
    def __init__(self):
        super().__init__()
        self.__secret = "Override attempt"

obj = Child()

# Python mangled the names to safely separate them:
print(obj._Parent__secret)  # Prints: "Hidden"
print(obj._Child__secret)   # Prints: "Override attempt"

# Trying to access the raw name will crash:
# print(obj.__secret)       # Raises AttributeError
```

##### **Unpack a Collection**

Unpacking is a convenient and readable feature that allows you to extract values from iterables (like lists, tuples, or strings) and assign them to multiple variables in a single line of code.

**Basic Unpacking**

```python
fruits = ["apple", "banana", "cherry"]
x, y, z = fruits
print(x, y, z, sep=("\n"))
```

**Extended unpacking**

If you have an iterable with many items but you only care about a few specific ones, you can use the asterisk (`*`) to "catch" the remaining items into a new list. This is called extended iterable unpacking.

```python
grades = [95, 85, 92, 78, 99]

# Grab the first grade and pack the rest into a list
first, *rest = grades
print(first, rest)

# It can also be used to grab the last item
*previous, last = grades
print(previous, [last])

# Or grab the first and last while packing the middle
first, *middle, last = grades
print(first, middle, last)
```

**Dictionary Unpacking**

While the single asterisk (`*`) unpacks the keys of a dictionary, the double asterisk (`**`) unpacks the *key-value* pairs.

```python
dict1 = {
    "a": 1,
    "b": 2
}
dict2 = dict(c = 3, d = 4)

print("Unpacked keys:", *dict1, *dict2)

merged = {**dict1, **dict2}
print("Unpacked and merged key-value:", merged)
```

*Note: If there are overlapping keys, the value from the dictionary unpacked last will overwrite the earlier one.*

```python
fifth_year = {
    "name": ["Oputa Chidubem", "Izuagie Itsoghena"],
    "CGPA": [4.0, 4.3]
}
fourth_year = {
    "name": ["Agene Philip", "Obot Silas"],
    "CGPA": [2.0, 3.1]
}
final_year = {**fifth_year, **fourth_year}
print(final_year)
```

**Unpacking in Function Arguments**

```python
def print_user(name, age, city):
    print(f"{name} is {age} and lives in {city}")

user_list = ["Bob", 30, "New York"]
user_dict = {"name": "Charlie", "age": 28, "city": "London"}

# Unpacking a list into positional arguments
print_user(*user_list)

# Unpacking a dictionary into keyword arguments
print_user(**user_dict)
```

```python
x = y = z = "Hello Wold!"

print(x, y, z)
print(x is y is z)  # compares their identities (memory addresses)
print(x == y == z)  # compares the values of two objects
```

The second statement printed is `True` because if you assign one variable to another, they point to the exact same memory slot.

Note: Python interprets both statements using a feature called **operator chaining**.Instead of evaluating left-to-right into a single boolean value first, Python automatically splits chained comparisons using an implicit `and` operator. It also evaluates the middle variable (y) only once.Here is exactly how Python expands and evaluates each line:
1. `print(x is y is z)`

Python rewrites this statement internally as:

`print((x is y) and (y is z))`
- Step 1: It checks if x and y point to the exact same memory address.
- Step 2: If x is y is False, Python stops immediately (short-circuits) and returns False.
- Step 3: If x is y is True, it then checks if y and z point to the same memory address.
- Final Result: It returns True only if all three variables point to the exact same object in memory.

2. `print(x == y == z)`

Python rewrites this statement internally as:

`print((x == y) and (y == z))`
- Step 1: It checks if the value of x is equal to the value of y.
- Step 2: If x == y is False, it short-circuits and returns False.
- Step 3: If x == y is True, it evaluates whether the value of y matches the value of z.
- Final Result: It returns True if all three variables hold identical data values, regardless of where they are stored in memory.

#### **Global Variables**

These are variables created outside of a function (as in all of the examples above). These variables can be used by everyone, both inside of functions and outside.

```python
x = "awesome"

def my_func():
    print("Python is " + x)

my_func()
```

If you create a variable with the same name inside a function, this variable will be local, and can only be used inside the function. The global variable with the same name will remain as it was (global and with the original value).

```python
def my_func():
    x = "fantastic"
    print("Python is " + x)

my_func()

print("Python is " + x)
```

Normally, when you create a variable inside a function, that variable is local, and can only be used inside that function. To create a global variable inside a function, you can use the `global` keyword.

```python
def my_func():
    global x
    x = "fantastic"
    print("Python is " + x)

my_func()

print("Python is " + x)
```

```python
def calculate_metrics(length, width):
    
    global height
    height = 10
    area = length * width
    volume = area * height

    # Inspect all local variables at once
    print(locals())

    return volume
    
calculate_metrics(5, 4)
```

**The Warlus Operator**

The walrus operator, introduced in Python 3.8 (PEP 572), lets you assign a value to a variable as part of an expression, rather than as a separate statement.

Unlike normal assignment `=`, which is a statement and doesn't return a value, `:=` is an expression, meaning it evaluates to the value being assigned, so you can use it inline inside other expressions (like `if`, `while`, list comprehensions, etc.).

Common use cases:

Avoiding duplicate computation in `while` loops

```python
# Without warlus
line = input("Enter text: ")

while line != "quit":
    print(f"You said: {line}")
    line = input("Enter text: ")  # duplicated

# With warlus
while (line := input("Enter text: ")) != "quit":
    print(f"You said: {line}")
```

**List Comprehension**

Syntax
> newlist = [expression for item in iterable if condition == True]

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

newlist = [x for x in fruits if "a" in x]

print(newlist) 
```

```python
newlist = [x for x in fruits if x != "apple"]
print(newlist)
```

```python
newlist = ["orange" if x == "banana" else x for x in fruits]
print(newlist)
```

```python
newlist = [x for x in range(10)]
print(newlist)
```

```python
newlist = [x for x in range(10) if x < 5]
print(newlist)
```

##### **Join Sets**

**Union and Update**

There are several ways to join two or more sets in Python.

The `union()` and `update()` methods joins all items from both sets.

The `union()` method returns a new set with all items from both sets.

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = set1.union(set2) # or set1 | set2
print(set3)

set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}
myset = set1 | set2 | set3 | set4 # unsorported operand when iterable is not a set
print(myset)

x = {"a", "b", "c"}
y = (1, 2, 3)
z = x.union(y) # joining a set with a tuple
print(z) 
```

The `update()` method inserts all items from one set into another. It changes the original set, and does not return a new set.

```python
set1 = {"a", "b" , "c"}
set2 = {1, 2, 3}
set1.update(set2) # or set1 & set2
print(set1) 
```

**Intersection**

The `intersection()` method will return a new set, that only contains the items that are present in both sets. In short, it keeps only the duplicates.

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set3 = set1.intersection(set2)
print(set3) 
```

The `intersection_update()` method will also keep ONLY the duplicates, but it will change the original set instead of returning a new set.

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set1.intersection_update(set2)
print(set1)

set1 = {"apple", 1,  "banana", 0, "cherry"}
set2 = {False, "google", 1, "apple", 2, True}
set3 = set1.intersection(set2)
print(set3)
```

**Difference**

The `difference()` method will return a new set that will contain only the items from the first set that are not present in the other set.

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set3 = set1.difference(set2) # or set1 - set2
print(set3) 
```

The `difference_update()` method will keep the items from the first set that are not in the other set, but it will change the original set instead of returning a new set.

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set1.difference_update(set2)
print(set1) 
```

**Symmetric Differences**

The `symmetric_difference()` method will keep only the elements that are NOT present in both sets.

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.symmetric_difference(set2) # or set1 ^ set2

print(set3) 
```

The `symmetric_difference_update()` method will also keep all but the duplicates, but it will change the original set instead of returning a new set.

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set1.symmetric_difference_update(set2)
print(set1) 
```
