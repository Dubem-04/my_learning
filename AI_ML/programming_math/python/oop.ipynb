#### What is an Object?

An object is a thing in memory with an identity, a type, and some state, and its type determines the behavior (methods) it supports. A class defines what its objects look like and can do. An instance is one object created from it.

An object has three things:

- Identity: A unique ID that never changes
- Type    : What kind of object it is which determines it's behaviour
- State   : The data it holds

```python
x = [1, 2, 3]     # State
print(id(x))      # Identity
print(type(x))    # Type
print(x)
```

```python
list_methods = [method for method in dir(x) if not method.startswith("__")]
print(list_methods)
```

Variable are merely name tags that point to an object stored in memory.

```python
a = [1, 2, 3]      # the state here ([1, 2, 3]) has an id that the variable "a" points to
b = a              # b is a second name for the SAME object
b.append(4)
print(a)
print(a is b)

c = a.copy()       # a new, separate object with equal value
print(c == a)
print(c is a)
```

In Python, everything is an object

```python
# The information within the type parenthesis are just instances of their respective classes
print(type(5))
print(type("hi"))
print(type(len))
print(type(print))
print(type(int))
print((5).bit_length())
```

Because functions and classes are objects, you can store them in variables, pass them as arguments, and return them from other functions.

Mutable vs Immutable Objects:

- Mutable: It's state can change in-place (`list`, `dict`, `set` and most instances of calsses i write)
- Immutable: It's state can't change after creation (`str`, `int`, `tuple`, `frozenset`). Modifying one ends up creating a new object.

```python
s = "abc"
print(id(s))
s += "d"           # creates a NEW string
print(id(s))       # different id
print()

nums = [1, 2]
print(id(nums))
nums.append(3)     # modified in place
print(id(nums))    # same id
```

A **public interface** is the set of all methods provided by a class, together with a description of their behaviour. When working with an object of a class, all I need to know is the public interface (which methods I can apply and what these methods do).

The process of providing a public interface while hiding the implementation details is called **encapsulation**. It enables changes in the implementation of a class without affecting the users of that class.

#### Implementing a simple class

```python
class Counter:

    def __init__(self):
        """Counter constructor"""
        self._strokes = ""

    def reset(self):
        """Resets the value of the counter to zero"""
        self._strokes = ""

    def click(self):
        """Increments the value of the counter by 1"""
        self._strokes = self._strokes + "|"

    def getValue(self):
        """Gets the current value of the counter"""
        return self._strokes
```

An **instance variable** is a storage location present in each instance of a class. They must be referenced within a method using the *self* parameter

The **self** parameter represents to the specific instance of the class on which a method is being called. It acts as a reference that allows the method to access and modify the object's unique attributes and methods.

```python
tally = Counter()

tally.reset()
tally.click()
tally.click()

result = tally.getValue()
print(f"Value: {result}")

tally.click()
result = tally.getValue()
print(f"Value: {result}")

tally_methods = [method for method in dir(tally) if not method.startswith("__")]
print(tally_methods)
```

#### Specifying the Public Interface of a Class

```python
class CashRegister:

    def addItem(self, price):
        """Adds an item to the cash register"""
        pass

    def getTotal(self):
        """Gets the price of all items in the current sale"""
        pass

    def getCount(self):
        """Gets the number of items in the current sale"""
        pass

    def getDollars(self):
        """Displays the amount of the total sales in dollar"""
        pass

    def clear(self):
        """Clears the item count and the total"""
        pass
```

A **mutator method** modifies the object on which it operators.

- The `CashRegister` class has two mutators: `addItem` and `clear`. After you call either of  these methods, the object has changes.

An **accessor method** queries the object for some information without changing 
it.
- The CashRegister class has two accessors: `getTotal` and `getCount`.

#### Designing the Data Representation

All instance variables should be private and most methods should be public. Although most 
object-oriented languages provide a mechanism to explicitly hide or protect private members from outside access, Python does not. Instead, the designer of a class has to indicate which instance variables and methods are supposed to be private. It’s then the responsibility of the class user not to violate the privacy.

It is common practice among Python programmers to use names that begin with a single 
underscore for private instance variables and methods. The single underscore serves as a flag to the class user that those members are private. You then must trust that the class user will not attempt to access these items directly. This technique is recognized by documentation generator tools that flag private instance variables and methods in the documentation.

You should always use encapsulation, in which all instance variables are private and are 
only manipulated with methods.
 
Typically, methods are public. However, sometimes you have a method that is used only as 
a helper method by other methods. In that case, you should make the helper method private 
by using a name that begins with a single underscore
