#### What is an Object?

An object is a thing in memory with an identity, a type, and some state. It's type determines the behavior (methods) it supports. A class defines what its objects look like and can do. An instance is one object created from it.

An object has three things:

- Identity: A unique ID that never changes
- Type    : What kind of object it is which determines it's behaviour
- State   : The data it holds

````python
x = [1, 2, 3]     # State
print(id(x))      # Identity
print(type(x))    # Type
print(x)
````

````python
list_methods = [method for method in dir(x) if not method.startswith("__")]
print(list_methods)
````

Variable are merely name tags that point to an object stored in memory.

````python
a = [1, 2, 3]      # the state here ([1, 2, 3]) has an id that the variable "a" points to
b = a              # b is a second name for the SAME object
b.append(4)
print(a)
print(a is b)

c = a.copy()       # a new, separate object with equal value
print(c == a)
print(c is a)
````

In Python, everything is an object

````python
# The information within the type parenthesis are just instances of their respective classes
print(type(5))
print(type("hi"))
print(type(len))
print(type(print))
print(type(int))
print((5).bit_length())
````

Because functions and classes are objects, you can store them in variables, pass them as arguments, and return them from other functions.

````python
def name(name):
    return name

me = name("Chidubem")

def greeting():
    return f"Hello {me}"

greeting()
````

Mutable vs Immutable Objects:

- Mutable: It's state can change in-place (`list`, `dict`, `set` and most instances of calsses I write)
- Immutable: It's state can't change after creation (`str`, `int`, `tuple`, `frozenset`). Modifying one ends up creating a new object.

````python
s = "abc"
print(id(s))
s += "d"           # creates a NEW string
print(id(s))       # different id
print()

nums = [1, 2]
print(id(nums))
nums.append(3)     # modified in place
print(id(nums))    # same id
````

A **public interface** is the set of all methods provided by a class, together with a description of their behaviour. When working with an object of a class, all I need to know is the public interface (which methods I can apply and what these methods do).

The process of providing a public interface while hiding the implementation details is called **encapsulation**. It enables changes in the implementation of a class without affecting the users of that class.

#### Implementing a simple class

The class below is a class that models a *tally counter*, a device used to count people.

Whenever the operator pushes a button, the counter value advances by one.

````python
class Counter:
    """Models a tally counter whose value can be incremented, viewed, or reset."""
    def click(self):
        """Advances the value of this counter by 1."""
        self._value += 1

    def getValue(self):
        """Gets the current value of this counter."""
        return self._value
    
    def reset(self):
        """Resets the value of this counter to 0."""
        self._value = 0
````

One thing wrong with the code above is that the `reset` method has to be called first in order to initialize the instance variable `_value`.

````python
tally = Counter()

try:
    print(tally.getValue())
except AttributeError:
    print("'Counter' object has no attribute '_value'")
````

An **instance variable** is a storage location present in each instance of a class. They must be referenced within a method using the *self* parameter. It should only be accessed by methods of its own class.

The **self** parameter represents to the specific instance of the class on which a method is being called. It acts as a reference that allows the method to access and modify the object's unique attributes and methods. Below, `tally` is the variable pointing to a specific instance of `Counter`.

````python
tally = Counter()

tally.reset()
tally.click()
tally.click()

result = tally.getValue()
print(f"Value: {result}")

tally.click()
result = tally.getValue()
print(f"Value: {result}")

tally_methods = [method for method in dir(tally) if not method.startswith("__")]
print(tally_methods)
````

#### Specifying the Public Interface of a Class

This is visualized using a simple example of an object that simulates a cash register.

A cashier who rings up a sale presses a key to start the sale, then rings up each item. A display shows the amount owed as well as the total number of items purchased. In our simulation, we want to call the following methods on a cash register object:
- Add the price of an item.
- Get the total amount owed, and the count of items purchased. 
- Clear the cash register to start a new sale.

````python
class CashRegister:
    """A simulated cash register that tracks the item count and the total amount due."""

    def __init__(self):
        self._cart = []

    def addItem(self, price: float) -> list:
        """Adds an item to this cash register"""
        self._cart.append(price)

    def getTotal(self) -> float:
        """Gets the price of all items in the current sale"""
        return sum(self._cart)

    def getCount(self) -> int:
        """Gets the number of items in the current sale"""
        return len(self._cart)

    def clear(self) -> list:
        """Clears the item count and the total"""
        return self._cart.clear()
````

The method definitions and comments make up the public interface of the class while the data and the method bodies make up the private implementation of the class. 

The methods of a class can be classified as either a **mutator** or an **accessor**.

A **mutator method** modifies the object it operates on.
- The `CashRegister` class has two mutators: `addItem` and `clear`. After you call either of  these methods, the object changes.

An **accessor method** queries the object for some information without changing it.
- The CashRegister class has two accessors: `getTotal` and `getCount`.

#### Constructors

A **constructor** defines and initializes the instance variables of an object and is called automatically when an object is created. Thus, giving an object a personality upon creation.

*Note: The **self** parameter in the constructor muct be the first parameter to allow for the unique personalities given to each instance (object) of the class*

````python
def __init__(self):
    self._itemCount = 0
    self.total_Price = 0
````

Python allows you to define only one constructor per class. But you can a define constructor with default argument values to provide different ways of creating a class.

````python
class BankAccount:

    def __init__(self, initial_balance=0):
        self._balance = initial_balance

myAccount = BankAccount()
print(myAccount._balance)

myAccount = BankAccount("$1T")
print(myAccount._balance)
````
