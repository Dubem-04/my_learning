#### What is an Object?

An object is a thing in memory with an identity, a type, and some state. It's type determines the behavior (methods) it supports. A class defines what its objects look like and can do. An instance is one object created from it.

An object has three things:

- Identity: A unique ID that never changes
- Type    : What kind of object it is which determines it's behaviour
- State   : The data it holds

````python
x = [1, 2, 3]     # State
print(id(x))      # Identity
print(type(x))    # Type
print(x)
````

````python
list_methods = [method for method in dir(x) if not method.startswith("__")]
print(list_methods)
````

Variable are merely name tags that point to an object stored in memory.

````python
a = [1, 2, 3]      # the state here ([1, 2, 3]) has an id that the variable "a" points to
b = a              # b is a second name for the SAME object
b.append(4)
print(a)
print(a is b)

c = a.copy()       # a new, separate object with equal value
print(c == a)
print(c is a)
````

````python
# The information within the type parenthesis are just instances of their respective classes
print(type(5))
print(type("hi"))
print(type(len))
print(type(print))
print(type(int))
print((5).bit_length())
````

Mutable vs Immutable Objects:

- Mutable: It's state can change in-place (`list`, `dict`, `set` and most instances of calsses I write)
- Immutable: It's state can't change after creation (`str`, `int`, `tuple`, `frozenset`). Modifying one ends up creating a new object.

````python
s = "abc"
print(id(s))
s += "d"           # creates a NEW string
print(id(s))       # different id
print()

nums = [1, 2]
print(id(nums))
nums.append(3)     # modified in place
print(id(nums))    # same id
````

#### Implementing a simple class

The class below is a class that models a *tally counter*, a device used to count people.

Whenever the operator pushes a button, the counter value advances by one.

````python
class Counter:
    """Models a tally counter whose value can be incremented, viewed, or reset."""
    def click(self):
        """Advances the value of this counter by 1."""
        self._value += 1

    def getValue(self):
        """Gets the current value of this counter."""
        return self._value
    
    def reset(self):
        """Resets the value of this counter to 0."""
        self._value = 0
````

One thing wrong with the code above is that the `reset` method has to be called first in order to initialize the instance variable `_value`.

````python
tally = Counter()

try:
    print(tally.getValue())
except AttributeError:
    print("'Counter' object has no attribute '_value'")
````

An **instance variable** is a storage location present in each instance of a class. They must be referenced within a method using the *self* parameter. It should only be accessed by methods of its own class.

The **self** parameter represents to the specific instance of the class on which a method is being called. It acts as a reference that allows the method to access and modify the object's unique attributes and methods. Below, `tally` is the variable pointing to a specific instance of `Counter`.

````python
tally = Counter()

tally.reset()
tally.click()
tally.click()

result = tally.getValue()
print(f"Value: {result}")

tally.click()
result = tally.getValue()
print(f"Value: {result}")

tally_methods = [method for method in dir(tally) if not method.startswith("__")]
print(tally_methods)
````

#### Specifying the Public Interface of a Class

This is visualized using a simple example of an object that simulates a cash register.

A cashier who rings up a sale presses a key to start the sale, then rings up each item. A display shows the amount owed as well as the total number of items purchased. In our simulation, we want to call the following methods on a cash register object:
- Add the price of an item.
- Get the total amount owed, and the count of items purchased. 
- Clear the cash register to start a new sale.

````python
class CashRegister:
    """A simulated cash register that tracks the item count and the total amount due."""

    def __init__(self):
        self._cart = []

    def addItem(self, price: float) -> list:
        """Adds an item to this cash register"""
        self._cart.append(price)

    def getTotal(self) -> float:
        """Gets the price of all items in the current sale"""
        return sum(self._cart)

    def getCount(self) -> int:
        """Gets the number of items in the current sale"""
        return len(self._cart)

    def clear(self) -> list:
        """Clears the item count and the total"""
        return self._cart.clear()
````

The method definitions and comments make up the public interface of the class while the data and the method bodies make up the private implementation of the class. 

The methods of a class can be classified as either a **mutator** or an **accessor**.

A **mutator method** modifies the object it operates on.
- The `CashRegister` class has two mutators: `addItem` and `clear`. After you call either of  these methods, the object changes.

An **accessor method** queries the object for some information without changing it.
- The CashRegister class has two accessors: `getTotal` and `getCount`.

#### Implementing Methods

Implementing methods is very similar to the implementation of functions with the exception that methods can access the instance variables of the object.

As with constructors, methods must include the `self` parameter variable, and it must be listed first.

When one method needs to call another method on the same object, you invoke the method on the `self` parameter. Suppose we want to provide a `CashRegister` method that adds multiple instances of the same item. An easy way to implement this method is to repeatedly call the `addItem` method:

````python
class CashRegister:
    """A simulated cash register that tracks the item count and the total amount due."""

    def __init__(self):
        self._cart = []

    def addItem(self, price: float) -> list:
        """Adds an item to this cash register"""
        self._cart.append(price)

    def addItems(self, quantity: int, price: float) -> list:
        """Adds multiple instances of an item"""
        for _ in range(quantity):
            self.addItem(price)

    def getTotal(self) -> float:
        """Gets the price of all items in the current sale"""
        return sum(self._cart)

    def getCount(self) -> int:
        """Gets the number of items in the current sale"""
        return len(self._cart)

    def clear(self) -> list:
        """Clears the item count and the total"""
        return self._cart.clear()
````

````python
register1 = CashRegister()

register1.addItems(10, 450.00)
print(register1.getCount())
print(register1.getTotal())
````

````python
class CashRegister:
    """A simulated cash register that tracks the item count and the total amount due."""

    def __init__(self):
        self._cart = []

    def addItem(self, price: float) -> list:
        """Adds an item to this cash register"""
        self._cart.append(price)

    def addItems(self, quantity: int, price: float) -> list:
        """Adds multiple instances of an item"""
        for _ in range(quantity):
            self.addItem(price)

    def getTotal(self) -> float:
        """Gets the price of all items in the current sale"""
        return sum(self._cart)

    def getCount(self) -> int:
        """Gets the number of items in the current sale"""
        return len(self._cart)

    def getDollar(self) -> int:
        """Rounds the total value to the nearest dollar"""
        return round(self.getTotal())

    def giveChange(self, payment: float) -> float:
        if payment < self.getTotal():
            print("Insufficient Funds")
        else:
            balance = f"#{payment - self.getTotal()}"
            self.clear()
            return balance

    def clear(self) -> list:
        """Clears the item count and the total"""
        self._cart.clear()
````

**Class variables (static variables)**

There are times when a piece of data needs to be shared across every single object created from a class, a class variable enables that.

````python
class BankAccount:
    _lastAssignedNumber = 1000

    def __init__(self):
        self._balance = 0
        BankAccount._lastAssignedNumber += 1
        self._accountNumber = BankAccount._lastAssignedNumber

    def getAccountNumber(self):
        return self._accountNumber
````

````python
dubem = BankAccount()
mezue = BankAccount()

print(dubem.getAccountNumber())
print(mezue.getAccountNumber())
````

*Note: Never use a class variable for a mutable object (like a `list` or `dict`) if you expect instances to manage their own collections. Because they point to the exact same memory location, modifying the list on one object modifies it for everyone.*

````python
class Student:
    grades = []

alice = Student()
bob = Student()

alice.grades.append(95)
print(bob.grades) # Bob accidentally inherits Alice's grade!
````

Instead, use an instance variable instead so that Python initializes a brand-new, isolated list for every single student.

````python
class Student:
    def __init__(self):
        self.grades = [] # Instance variable list (Isolated)
````

````python
class Menu:
    """A menu that is displayed in a terminal window"""

    def __init__(self):
        """Constructs a menu with no options"""
        self._options = []

    def addOption(self, option):
        """Adds an option to the end of this menu"""
        self._options.append(option)
    
    def getInput(self):
        """
        Displays the options numbered starting with 1, and prompts the user for input.
        Repeats until a valid input is entered
        """

        while True:
            for i in range(len(self._options)):
                print(f"{i + 1}. {self._options[i]}")

            userChoice = int(input())
            if userChoice >=1 and userChoice < len(self._options):
                break

        return userChoice
````

````python
mainMenu = Menu()
mainMenu.addOption("Open new account")
mainMenu.addOption("Log into existing account")
mainMenu.addOption("Help")
mainMenu.addOption("Quit")
choice = mainMenu.getInput()
print("Input:", choice)
````

````python
class BankAccount:
    """Simulates a real-world banking system"""
    def __init__(self, initialBalance=0):
        """Constructs an object with an empty initial balance"""
        self._balance = initialBalance

    def deposit(self, amount):
        """Adds a deposit amount to the current balance"""
        self._balance += amount

    def withdraw(self, amount):
        """Subtracts withdrawal amount from the current balance"""
        PENALTY = 10.0
        if amount > self._balance:
            self._balance -= PENALTY
        else:
            self._balance -= amount

    def addInterest(self, rate):
        """Adds a variable interest every 30 days"""
        interest = self._balance * (rate/100)
        self._balance += interest

    def getBalance(self):
        """Displays current balance after an activity on the account"""
        return self._balance
````
